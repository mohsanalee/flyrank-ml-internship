## 5. Why ML beats a fixed rule here

**Why ML beats a fixed rule**

A fixed rule would rely on one or two manually chosen thresholds, such as flagging every page with low traffic or low clicks. That would miss the fact that content performance depends on several signals at the same time, including impressions, clicks, engagement, content age, content type, and other page-level characteristics. These relationships can also vary across clients and content items. ML is useful when the pattern is real but too messy to describe with a small set of hand-written rules. The model's output would help editors prioritize which pages to review first rather than automatically making the content decision for them.
